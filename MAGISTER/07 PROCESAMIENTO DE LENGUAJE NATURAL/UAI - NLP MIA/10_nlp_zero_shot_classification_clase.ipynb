```python
#Importar Librerías
import pprint
from transformers import pipeline
```

```python
#Labels para Clasificación
candidate_labels = ["politics",
                    "sports",
                    "culture"]
```

```python
#Crear Pipeline de Clasificación
classifier = pipeline(
    task="zero-shot-classification",
    device=0,
    model="facebook/bart-large-mnli"
)
```

```python
#Ejemplo 1
text_1 = "Olivia Newton-John, who has died aged 73, began her career as the sweetheart of country-flavoured pop."
pprint.pprint(classifier(text_1, candidate_labels))
```

```python
#Ejemplo 2
text_2 = "The Commonwealth Games ended on Monday with a familiar look"
pprint.pprint(classifier(text_2, candidate_labels))
```

```python
#Ejemplo 3
text_3 = "Boris Johnson rules out immediate cost-of-living measures"
pprint.pprint(classifier(text_3, candidate_labels))
```

```python
#Ejemplo 4 (Multi-Label)
text_3 = "Boris Johnson rules out immediate cost-of-living measures"
pprint.pprint(classifier(text_3, candidate_labels, multi_label=True))
```

```python
#Ejemplo 5
text_4 = "Robert Lewandowski Participates in “Reds Against Racism” Roundtable Through Former Club"
pprint.pprint(classifier(text_4, candidate_labels))
```
