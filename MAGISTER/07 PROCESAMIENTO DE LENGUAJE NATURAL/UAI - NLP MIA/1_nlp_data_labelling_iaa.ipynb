### Data Labelling & Inter-Annotation Agreement

```python
#Importar Librerías
import pandas as pd
from sklearn.metrics import f1_score,confusion_matrix,cohen_kappa_score
```

```python
#Importar Archivo JSON
data = pd.read_csv('/content/data_labelling_sentiment_analysis_outcome.csv', encoding='latin-1')
data
```

```python
#Calcular Inter-Annotation Agreement (IAA - F1 Score)
average='weighted'
iaa_f1_score = f1_score(data.annotator_1, data.annotator_2, average=average)
print(f'F1 Score({average}): {round(iaa_f1_score,2)}')
```

```python
#Calcular Inter-Annotation Agreement (IAA - F1 Score)
iaa_kappa = cohen_kappa_score(data.annotator_1, data.annotator_2)
print(f"Cohen's kappa: {iaa_kappa}")
```
