### Pre-Processing, Lemmatization & POS Tagging (Inglés)

```python
#Importar Librerías
import spacy
```

```python
#Cargar Textos
texts = ["Bank of England warns the UK will fall into recession this year",
         "Brittney Griner: US basketball star jailed for nine years on drug charges",
         "China fires missiles near Taiwan after Pelosi visit",
         "Iceland volcano was so beautiful, I had a little cry",
         "Premier League: A season like no other, but some things won't change"]
```

```python
#Descargar Pipeline (English)
!python -m spacy download en_core_web_sm -q
```

```python
#Cargar Pipeline
nlp = spacy.load("en_core_web_sm")
```

```python
#Crear Doc Objects
docs = []
for text in texts:
  docs.append(nlp(text))
```

```python
#POS Tagging (Nivel General & Detalle)
for token in docs[0]:
  print(token.text,"---",token.pos_,"---",token.tag_)
```

```python
#Significado POS Tags
print(spacy.explain("PROPN"))
print(spacy.explain("NNP"))
```

```python
#Visualización POS Tags (Syntactic Dependencies)
from spacy import displacy
displacy.render(docs[0],style='dep',jupyter=True)
```

```python
#Limpieza Textos (Tokenización, Lemmatización, Lowercasing, Stopwords & Signos de Puntuación)
cleaned_docs = []
for doc in docs:
  cleaned_docs.append([token.lemma_.lower() for token in doc if not token.is_stop and not token.is_punct])
```

```python
#Revisar Resultado Limpieza
print("Original Text:", [token.text for token in docs[0]])
print("Post Processing Tokens:", cleaned_docs[0])
```
