```python
#Almacenar Labels en Test Set
spam_test = [labels for email, labels in test_data]
```

```python
#Función para Obtener Respuesta
def get_completion(prompt, model="gpt-3.5-turbo"):
    messages = [{"role": "user", "content": prompt}]
    response = openai.ChatCompletion.create(
        model=model,
        messages=messages,
        temperature=0.0,
    )
    return response.choices[0].message["content"]
```

# Promt Engineering (Zero Shot)



```python
#Prompt - Zero Shot
pred_spam_zero = []
for email, labels in test_data:

  zero_prompt = f"""
  You are a cybersecurity expert, and you will be presented with emails and asked whether they are spam.
  Format your response as a binary variable, using 'spam' if the email is spam and 'ham' otherwise.
  Triple backticks delimit each email.

  Email: '''{email}'''
  """

  response = get_completion(zero_prompt)
  pred_spam_zero.append(response)
```

```python
#Print Pred Spam
pred_spam_zero
```

```python
#Calcular F1 Score (Spam Class)
f1_score(spam_test, pred_spam_zero, pos_label='spam')
```

# Promt Engineering (Few Shot)

```python
#Prompt - Few Shot
pred_spam_few = []
for email, labels in test_data:

  few_prompt = f"""
  You are a cybersecurity expert, and you will be presented with emails and asked whether they are spam.
  Format your response as a binary variable, using 'spam' if the email is spam and 'ham' otherwise.
  Triple backticks delimit each email.

  Thanks for your subscription to Ringtone UK your mobile will be charged Â£5/month Please confirm by replying YES or NO. If you reply NO you will not be charged // spam
  Did you hear about the new 'Divorce Barbie'? It comes with all of Ken's stuff! // spam
  Thanks a lot for your wishes on my birthday. Thanks you for making my birthday truly memorable // ham
  Please call our customer service representative on 0800 169 6031 between 10am-9pm as you have WON a guaranteed Â£1000 cash or Â£5000 prize! // spam

  Email: '''{email}'''
  """

  response = get_completion(few_prompt)
  pred_spam_few.append(response)
```

```python
#Print Pred Spam
pred_spam_few
```

```python
#Calcular F1 Score (Spam Class)
f1_score(spam_test, pred_spam_few, pos_label='spam')
```
