### Data Labelling & Inter-Annotation Agreement

```python
#Importar Librerías
import pandas as pd
from sklearn.metrics import f1_score,confusion_matrix,cohen_kappa_score
```

```python
#Importar Archivo JSON
data = pd.read_csv('/content/data_labelling_sentiment_analysis_outcome.csv', encoding='latin-1')
data
```

```python
#Calcular Inter-Annotation Agreement (IAA - F1 Score)
iaa_f1_score = f1_score(data.annotator_1, data.annotator_2, average='weighted')
print("F1 Score (IAA):", round(iaa_f1_score,3))
```

```python
#Calcular Inter-Annotation Agreement (IAA - Cohen's Kappa Score)
iaa_cohen_kappa = cohen_kappa_score(data.annotator_1, data.annotator_2)
print("Cohen's Kappa Score (IAA):", round(iaa_cohen_kappa,3))
```
