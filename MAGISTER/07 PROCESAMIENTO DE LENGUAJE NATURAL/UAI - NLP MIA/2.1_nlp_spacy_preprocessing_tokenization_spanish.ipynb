### Pre-Processing, Lemmatization & POS Tagging (Español)

```python
#Importar Librerías
import spacy
```

```python
#Crear un Texto
text = "Orlando Bloom sorprendió con radical cambio de look: estaría en campaña para agrandar su familia con Katy Perry"
```

```python
#Descargar Pipeline (Spanish)
!python -m spacy download es_core_news_sm -q
```

```python
#Cargar Pipeline Model
nlp = spacy.load("es_core_news_sm")
```

```python
#Crear Doc Object
doc = nlp(text)
```

```python
#Tokenización e Identificación de Stopwords & Signos de Puntuación
for token in doc:
  print(token.text,"---", token.is_stop,"---",token.is_punct)
```

```python
#Lemmatization
for token in doc:
  print(token.text,"---",token.lemma_)
```

```python
#Otros atributos de los tokens (Reconocimiento de Emails)
email_text = "Mis direcciones de correo son erick.mendez@edu.uai.cl y erick.mendezguzman@manchester.ac.uk"
email_doc = nlp(email_text)
for token in email_doc:
  if token.like_email:
    print(token.text)
```

```python
#POS Tagging (Universal POS Tags)
for token in doc:
  print(token.text,"---",token.pos_)
```

```python
#Significado POS Tags
spacy.explain("PROPN")
```

```python
#Visualización POS Tags (Syntactic Dependencies)
from spacy import displacy
displacy.render(doc,style='dep',jupyter=True)
```
