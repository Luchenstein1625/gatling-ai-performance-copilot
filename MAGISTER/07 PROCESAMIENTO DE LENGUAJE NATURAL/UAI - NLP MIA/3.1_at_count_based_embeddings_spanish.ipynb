# Pre-Processing

```python
#Importar Librerías
import spacy
import pandas as pd
import numpy as np
```

```python
#Cargar Data
data = [
        "Mi mamá ama las manzanas",
        "Katy Perry odia las manzanas.",
        "Orlando Bloom compró acciones en NotCo",
        "Mi hermana nació en Melipilla",
        "Katy Perry visitó Melipilla",
        "Los fundadores de NotCo nacieron en Melipilla",
        "No me gustó la comida en Melipilla",
        "Orlando Bloom compró una casa en Melipilla",
        "Mi postulación a NotCo fue rechazada",
        "Mi papá fue al colegio en Melipilla",
        "El que fue a Melipilla perdió su silla",
        "No me gustan los productos de NotCo",
        "Anoche hubieron protestas contra NotCo"
        ]
```

```python
#Descargar Pipeline (Spanish)
!python -m spacy download es_core_news_sm -q
```

```python
#Cargar Pipeline Model
nlp = spacy.load("es_core_news_sm")
```

```python
#Función para Pre-Processing + Tokenización
def get_tokens(text):
  doc = nlp(text)
  tokens = [token.lemma_.lower() for token in doc if token if not token.is_stop and not token.is_punct]
  return tokens
```

# Modelo BoW (Frequency)

```python
#Importar Librerías
from sklearn.feature_extraction.text import CountVectorizer
```

```python
#Crear Objeto CountVectorizer
count_vectorizer = CountVectorizer(lowercase=False,
                                   preprocessor=None,
                                   tokenizer=get_tokens)
```

```python
#Estadísticas Descriptivas
count_vectorizer.fit(data)
print("Vocabulario:", count_vectorizer.get_feature_names_out())
print("Tamaño Vocabulario:", len(count_vectorizer.get_feature_names_out()))
```

```python
#Generar Representación Vectorial
data_count_vectorized = count_vectorizer.transform(data)
print("Dimensionalidad Rep. Vectorial:", data_count_vectorized.shape)
```

```python
#Revisar Sparse Matrix
data_count_vectorized.todense()
```

# Latent Semantic Analysis (Reducción Dimensionalidad)

```python
#Importar Librerías
from sklearn.decomposition import TruncatedSVD
```

```python
#Parametrización & Ajuste Objeto SVD
n_components = 10
SVD = TruncatedSVD(n_components=n_components,
                   random_state=0)
SVD.fit(data_count_vectorized)
```

```python
#Resumen Varianza Explicada
print("Total Varianza Explicada:", round(SVD.explained_variance_ratio_.sum(),3))
print("Varianza Explicada por Componente:\n", SVD.explained_variance_ratio_)
```

```python
#Transformación Matriz Rep. Vectorial
data_count_lsa = SVD.transform(data_count_vectorized)
print("Dimensionalidad Rep. Vectorial:", data_count_lsa.shape)
```
