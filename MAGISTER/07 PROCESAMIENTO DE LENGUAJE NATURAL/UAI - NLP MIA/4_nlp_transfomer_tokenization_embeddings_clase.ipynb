```python
#Importar Librerías
from transformers import AutoTokenizer
from transformers import AutoModel
import torch
```

```python
#Cargar Tokenizadores
tokenizer_en = AutoTokenizer.from_pretrained("bert-base-uncased")
tokenizer_es = AutoTokenizer.from_pretrained("dccuchile/bert-base-spanish-wwm-uncased")
```

```python
#Cargar Ejemplos
sentence_en = "Natural Language Processing is fascinating."
sentence_es = "El procesamiento del lenguaje natural es fascinante."
```

```python
#Tokenización de Ejemplos
tokens_en = tokenizer_en.tokenize(sentence_en)
tokens_es = tokenizer_es.tokenize(sentence_es)
```

```python
#Revisar Tokenización
print("English Tokens:\n", tokens_en)
```

```python
#Tokenización de Ejemplos (IDs)
input_ids_en = tokenizer_en.encode(sentence_en, add_special_tokens=True)
input_ids_es = tokenizer_es.encode(sentence_es, add_special_tokens=True)
```

```python
#Revisar IDs
print("Token IDs (English):", input_ids_en)
```

# Word Embeddings

```python
#Cargar Modelos
model_en = AutoModel.from_pretrained("bert-base-uncased")
model_es = AutoModel.from_pretrained("dccuchile/bert-base-spanish-wwm-uncased")
```

```python
#Crear Input (Tensores PyTorch)
inputs_en = tokenizer_en(sentence_en, return_tensors="pt")
inputs_es = tokenizer_es(sentence_es, return_tensors="pt")
```

```python
#Forward Pass (Sin Cálculo Gradiente)
with torch.no_grad():
    outputs_en = model_en(**inputs_en)
    outputs_es = model_es(**inputs_es)
```

```python
#Revisar Word Embeddings (Token-Level)
print("Dimensionalidad Embeddings (Tokens):", outputs_en.last_hidden_state.shape)
```

```python
#Revisar Word Embeddings (Document-Level)
print("Dimensionalidad Embeddings (Documento):", outputs_en.pooler_output.shape)
```
