# Datos Faltantes o Missing Values

Un dato faltante consiste en valores no observados para alguna(s) variable(s), en algún(os) registro(s). Lo anterior es bastante sensible, en particular cuando estamos pensando en generar algún tipo de modelo o estudio, estos valores pueden generar problemas o influencias en nuestros proceso.

Las causas por las cuales se generan estos datos faltantes son variadas, pero las más típicas son:

1. **No ingresadas, ya sea intencionalmente o no:** La persona o maquina encargada de recolectar los datos, omitió algunos de manera intencionada o no (Ej. En un fraude se quiere omitir un registro para eliminar evidencia)
2. **No disponibles:** Si generando registros de individuos, pero dentro de la recolección existen individuos sin ese dato (Ej. Muestras de sangre para revisar colesterol, y tenemos un individuo sin el examen correspondiente)
3. **Errores:** Estos se pueden deber a multiples cosas, tanto errores humanos como de sistemas.

Los efectos como ya se mencionaron, pueden ser los siguientes:

1. **Problemas con los módulos utilizados** (No todos tienen tratamiento de datos faltantes)
2. **Datos distorcionados** (Al faltar registros, podemos no estar considerando parte importante de los datos)
3. **Performance de modelos** (Nuevamente no todos los modelos tienen tratamiento de estos datos)

En Python, los datos faltantes son representados con `NaN`, que significa "Not a Number". También podemos encontrar valores representados por `None`, donde la diferencia es la siguiente

- `None` representa un dato faltante, pero su tipo no es numéricos, esto es, que cualquier columna que contenga un `None` no puede ser de tipo numérico.
- `NaN` signfica "Not a Number", por lo que el registro con `NaN` no es un número en una columna que si es de tipo numérico (`int`, `float`).

Como ya mencionamos antes, existen diversas formas en que se pueden producir datos faltantes, y aquí tenemos algunas de esas formas de manera más ejemplificada:

- Error Humano
- Encuestados no quieren responder ciertas preguntas
- Encuestador no entiende la pregunta
- El valor obtenido es claramente erróneo, por tanto se elimina
- Sin tiempo para responder
- Registros perdidos debido a la falta de una gestión eficaz de la base de datos
- Eliminación intencional y omisión de la recopilación de los datos
- Participante abandona en medio del estudio
- Manipulación de terceros o bloqueo de la recopilación de datos
- Observaciones perdidas
- Bugs en programas
- Malfuncionamiento del sensor

De todas maneras, e independiente de qué causó los datos faltantes, podemos categorizar este tipo de errores en 3 tipos, como sigue:

- **Missing Completely at Random (MCAR):** Estos son valores faltantes completamente al azar, y se refiere a ellos de esta manera a aquellos en los que no hay razón para creer que los valores faltan debido a alguna razón sistemática. Cuando un dato faltante es clasificado en **MCAR**, entonces la falta de ese dato podría haber sido perfectamente cualquier otra.
    - **Ejemplo:** Si un sensor de calidad de aire le falla la comunicación con el servidor para registrar los datos, debido a la caída de conexión de su internet satelital debido al paso de un asteroide muy cerca del satelite.

- **Missing at Random (MAR):** Este tipo ocurre cuando cuando la probabilidad de ser dato faltante en una variable a lo largo de los registros, no es uniforme.
    - **Ejemplo:** Si vientos fuertes causan aveces el malfuncionamiento de un sensor y por ello, este no obtiene registros, entonces será más probable tener datos faltantes en día de viento fuerte que cuando no.
    
    
- **Missing not at Random (MNAR):** Este tipo, y tal como lo dice su nombre, no tiene nada de aleatorio, es decir, sabremos exactamente que observación tendrá un valor faltante.
    - **Ejemplo:** Si tenemos un sensor que ha sido manipulado para que no muestre su datos.

## Tratamiento de Missing Values

Existen distintos enfoques cuando nos enfrentamos a datos faltantes, en el sentido del tratamiento que debe darseles (Porque hay que hacer algo, o vivir con las consecuencias de no hacer nada). En este sentido, tenemos las siguientes opciones:

- Conservar los datos originales
- Eliminar las filas con datos faltantes
- Eliminar las variables con datos faltantes
- Imputar un valor

Si conservamos los datos originales, tal y como vienen, entonces tendremos que siempre tener cuidado con nuestros análisis, así como también con los futuros modelos que podremos utilizar, dado que no todos los modelos tienen tratamiento de datos faltantes.

Ahora para ejemplificar los otros métodos, consideremos el siguiente dataset

Respecto a este método, podemos comentar lo siguiente:

**Ventajas**:
- Extremadamente simple de aplicar.
- Si los missing values son completamente al azar, entonces al eliminarlos la distribución de la variable se conserva.

**Desventajas**:
- Si la cantidad de missing values es alta, una gran cantidad de data será eliminada.
- No aplicable para missing values que no sean completamente aleatorios.

Respecto al segundo método, de eliminar las variables que contienen datos faltantes, tendríamos lo siguiente:

```python
ej_data.dropna(axis= 1)
```

Para hacer esto, se deben tener las siguientes consideraciones:

- Se eliminan aquellas variables que no son claves para el proyecto a realizar.
- Generalmente se opta por este método cuando la cantidad de missing values supera el 25%


### Imputación de missing values

La imputación de datos es una técnica que nos permite no perder registros, cambiando los valores `NaN` por otro (que dependerán de la técnica que se utilice). En este sentido tenemos dos opciones que son relativamente comunes:

- **Imputación por medida de tendencia central** 
- **Imputación por regresión**

Imputación por Media o Mediana o Moda

Ejemplo:

```python
ej_data
```

```python
ej_data.Age.mean() ## La función mean, realiza el dropna de manera interna, por lo que no es necesario volver a hacerlo.
```

```python
ej_data.Age.median()
```

```python
ej_data.BloodGroup.mode()[0]
```

```python
#Imputamos con la media y la moda
ej_data.fillna({'Age':np.round(ej_data.Age.mean(),0), 'BloodGroup':ej_data.BloodGroup.mode()[0]})
```

```python
#Imputamos con la mediana y la moda
ej_data.fillna({'Age':np.round(ej_data.Age.median(),0), 'BloodGroup':ej_data.BloodGroup.mode()[0]})
```

- Se recomienda usar cuando la data faltante es al azar

#### Imputación por valor arbitrario

- Se escoge un valor cualquiera para reemplazar los missing values
- El valor NO debe pertenecer al dataset

Veamos ahora un ejemplo utilizando el dataset Titanic

```python
titanic_data_sub  = titanic_data[["survived", "pclass", "age", "fare"]]
```

Las estadísticas descriptivas nos ayudan a tener nociones de como se están comportando nuestros datos y que podemos esperar de ellos en relación a sus valores. En este sentido, las estadísticas básicas que se utilizan son:

- Medidas de tendencia central: Media, Mediana, Moda
- Medidas de dispersión: Rango, Varianza, Desviación estándar
- Medidas de forma: Skewness y Kurtosis
